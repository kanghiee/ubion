# 마케팅 채널별 유입수 및 유입당 결제율

```python
df3 = df1.groupby(['채널명'])['유입수','결제금액(+14일기여도추정)','결제수(+14일기여도추정)'].sum()
df3['결제금액(+14일기여도추정)'] = round(df3['결제금액(+14일기여도추정)'])
df3['결제금액(+14일기여도추정)'] = df3['결제금액(+14일기여도추정)'].astype(int)
df3.head(10)
```

```python
df3['유입당_결제율'] = df3['결제수(+14일기여도추정)'] / df3['유입수'] *100
df3['유입당_결제금액'] = df3['결제금액(+14일기여도추정)']/df3['유입수']
df3
```

```python
df3.to_csv('마케팅채널별_유입.csv')
```
