```python
# 라이브러리 불러오기
from selenium import webdriver 
# 웹 브라우저를 실행
driver = webdriver.Chrome()

# 웹 브라우저에 주소를 입력
driver.get('https://smartstore.naver.com/hu100/products/8959294188?NaPm=ct%3Dlz6i0bmw%7Cci%3D52af28d84723ca60e76544711828a9d6e7951ad6%7Ctr%3Dsls%7Csn%3D8315074%7Chk%3D06e913f4732f43d2858646a3ec2d9351ed37f06e')
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import pandas as pd
import re
re_reviews = driver.find_elements(By.CSS_SELECTOR,'ul._2ms2i3dD92 li.BnwL_cs1av._nlog_click._nlog_impression_element')
```

```python
#page1
all_reviews = []
for rr in re_reviews:
    content = rr.find_element(By.CSS_SELECTOR,'._3z6gI4oI6l ._1kMfD5ErZ6 span._2L3vDiadT9').text
    content = re.sub('[^#0-9a-zA-Zㄱ-힣 ]', "", content)
    all_reviews.append(content)
```

```python
#page2
all_reviews2 = []
for rr in re_reviews:
    content = rr.find_element(By.CSS_SELECTOR,'._3z6gI4oI6l ._1kMfD5ErZ6 span._2L3vDiadT9').text
    content = re.sub('[^#0-9a-zA-Zㄱ-힣 ]', "", content)
    all_reviews2.append(content)
```
