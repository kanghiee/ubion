```python
# 라이브러리 불러오기
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, NoSuchElementException, ElementClickInterceptedException
import pandas as pd
import time

# 웹 브라우저를 실행
driver = webdriver.Chrome()

# 웹 브라우저에 주소를 입력
driver.get('https://front.wemakeprice.com/open/cs/policies/thirdparty')

# WebDriverWait 객체 정의
wait = WebDriverWait(driver, 10)
all1 = []

while True:
    try:    
        lli = wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, 'body > div > div.content > div > div > div.company_list > ul > li')))
        for i in lli:
            all1.append(i.text)
        
        # "다음 페이지" 버튼 클릭
        next_button = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'body > div > div.content > div > div > div.paging_comm > a.ico.btn_next')))
        next_button.click()
        time.sleep(1)  # 페이지가 로드될 시간을 주기 위해 일시 정지

    except (TimeoutException, NoSuchElementException, ElementClickInterceptedException) as e:
        print("No more pages or an error occurred.")
        print(e)
        break

# 드라이버 종료
driver.quit()

# 수집된 데이터를 DataFrame으로 변환
df = pd.DataFrame(all1, columns=['Company Name'])

# DataFrame을 엑셀 파일로 저장
df.to_excel('company_names.xlsx', index=False)

print("Data has been saved to 'company_names.xlsx'")
```
