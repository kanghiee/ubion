### 결혼답례품 스마트스토어 리뷰 형태소 분석한 csv파일 확인
- 상품의 대한 칭찬, 배송이 빠르다, 포장에 대한 만족감의 리뷰내용이 많다

```python
import pandas as pd

key_list = pd.read_csv('/Users/kang_hiee/Documents/GitHub/ubion/python/최종프로젝트/키워드분석/review_frequecy.csv')
key_list.loc[key_list['Frequency']>=100]
```

### 형태소를 단어 단위로 다시 분리하여 좀 더 상세한 키워드 확인

```python
from collections import Counter
# 각 단어에 대해 빈도수를 다시 계산할 딕셔너리 생성
word_freq = Counter()

# 데이터프레임에서 형태소와 빈도수를 가져옵니다.
for index, row in key_list.iterrows():
    words = row['Word'].split()  # 형태소를 단어 단위로 분리
    freq = row['Frequency']
    for word in words:
        word_freq[word] += freq

# 데이터프레임으로 변환
df_word_freq = pd.DataFrame(word_freq.items(), columns=['Word', 'Frequency'])

df_word_freq = df_word_freq.sort_values(by=['Frequency'], ascending =False)
df_word_freq[:50]
```

### 메인 키워드, 맨앞 키워드, 맨뒤 키워드 지정

1. 메인키워드
- 메인 키워드로는 '호두강정', '호두정과', '수제호두강정','수제호두정과','호두강정답례품', '호두정과답례품','수제호두강정답례품','수제호두정과답례품'

```python
main = ['호두강정', '호두정과', '수제호두강정','수제호두정과','호두강정답례품', '호두정과답례품',
        '수제호두강정답례품','수제호두정과답례품']
```

2. 맨앞 키워드
- 위의 키워드 빈도수 데이터프레임을 참고하여 맨앞 키워드 선정

```python
first = ['고급','깔끔한','고급스러운','배송빠른','맛있는','이쁜','꼼꼼한',
         '센스있는','맞춤','최고의','만족스러운','가성비','적당한','정성스러운',
         '간편한','감동적인','가벼운','간단한','완벽한','소량','대량','특별한','세련된','빠른','알찬','야무진',
         '포장이 깔끔한','인기있는','실용적인']
```

3. 맨뒤 키워드
- 위의 키워드 빈도수 데이터프레임을 참고하여 맨뒤 키워드 선정

```python
last = ['패키지','추천','제품','상품','선물','준비','구성','포장',
        '스티커','배송','후기','리뷰','세트','문구','제작','주문','업체']
```

# 키워드 조합
- 메인키워드 : 1, 맨앞에 올 키워드 : 2, 맨 뒤에 올 키워드 :3
- 2+1, 1+3, 2+1+3 조합

```python
# 조합을 담을 리스트
combinations = []

# 조합 생성
for main_keyword in main:
    for first_keyword in first:
        # 2 + 1 조합
        combinations.append(f"{first_keyword} {main_keyword}")
    
    for last_keyword in last:
        # 1 + 3 조합
        combinations.append(f"{main_keyword} {last_keyword}")
    
    for first_keyword in first:
        for last_keyword in last:
            # 2 + 1 + 3 조합
            combinations.append(f"{first_keyword} {main_keyword} {last_keyword}")

# 결과 출력
for combo in combinations:
    print(combo)

# 필요 시 조합을 CSV 파일로 저장
df_combinations = pd.DataFrame(combinations, columns=['Combination'])
df_combinations.to_csv('메인키워드+서브키워드조합(결혼답례품).csv', index=False)
df_combinations.to_excel('메인키워드+서브키워드조합(결혼답례품).xlsx', index=False)
```
